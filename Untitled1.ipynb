```python
# --- Day 17: 3-Qubit Error Correction Code ---
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

# 1. Setup Circuit (3 Physical Qubits for 1 Logical Qubit)
qc = QuantumCircuit(3, 1) # Measure only the main qubit (q0)

# A. Prepare the State (The Data)
# Let's say we want to protect the state |1>
qc.x(0) 

qc.barrier()

# B. Encode (Spread the info)
# We use CNOTs to copy q0's state to q1 and q2.
# |1> becomes |111>
qc.cx(0, 1)
qc.cx(0, 2)

qc.barrier()

# C. The Attack (Simulated Noise)
# Let's Flip Qubit 0 (The worst case scenario!)
# Ideally, q0 should be 1. We flip it to 0.
print("Injecting Error on Qubit 0...")
qc.x(0) 

qc.barrier()

# D. Decode & Correct (The Magic)
# Step 1: Compute Parity (Syndrome)
# We check differences between neighbors
qc.cx(0, 1)
qc.cx(0, 2)

# Step 2: Majority Vote Correction (Toffoli Gate)
# If q1 and q2 are both 1 (meaning q0 was different from them), flip q0 back.
qc.ccx(1, 2, 0)

qc.barrier()

# E. Verify
# We measure q0. It SHOULD be 1 (restored).
# If correction failed, it would be 0 (broken).
qc.measure(0, 0)

# 2. Draw
print("Bit-Flip Correction Circuit:")
display(qc.draw(output='mpl'))

# 3. Simulate
sim = AerSimulator()
qc_t = transpile(qc, sim)
result = sim.run(qc_t, shots=1000).result()
counts = result.get_counts()

# 4. Analyze
print("\nFinal Result (Should be '1'):")
print(counts)
display(plot_histogram(counts))
```
